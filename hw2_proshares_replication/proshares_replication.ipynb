## Setup

* **Data:** `data/proshares_analysis_data.xlsx`, monthly data on financial indexes and ETFs from Aug 2011 through Sep 2025 (the case text says May 2025; the file we received runs through Sep 2025).
* **Format:** annualize mean (× 12), volatility (× $\sqrt{12}$), and Sharpe (× $\sqrt{12}$) where asked. Tail-risk statistics in 2.2 are **not** annualized.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm

pd.options.display.float_format = '{:,.4f}'.format
plt.rcParams['figure.figsize'] = (10, 5)

FREQ = 12                                   # monthly data
DATA_PATH = 'data/proshares_analysis_data.xlsx'
```

```python
sheets = pd.read_excel(DATA_PATH, sheet_name=None, index_col=0)
sheets.pop('descriptions', None)          # ticker -> security name, not a time series
for df in sheets.values():
    df.index = pd.to_datetime(df.index)
print('Sheets:', list(sheets))

hf      = sheets['hedge_fund_series']       # HFRI, ML series, HDG, QAI, ...
factors = sheets['merrill_factors']         # SPY + other Merrill style factors

print(f'Sample: {hf.index[0]:%Y-%m} to {hf.index[-1]:%Y-%m}  ({len(hf)} monthly obs)')
print('Hedge fund series:', list(hf.columns))
print('Merrill factors  :', list(factors.columns))
```

## 2.2 Tail Risk
> For the series in the "hedge fund series" tab, calculate the following statistics related to tail-risk.
> * Skewness
> * Excess Kurtosis (in excess of 3)
> * VaR (.05) - the fifth quantile of historic returns
> * CVaR (.05) - the mean of the returns at or below the fifth quantile
> * Maximum drawdown - include the dates of the max/min/recovery within the max drawdown period.
>
> There is no need to annualize any of these statistics.

```python
# Uses raw (total) returns, not excess: tail-risk stats describe the series' own return distribution, so no rf adjustment (unlike 2.1's Sharpe).
def max_drawdown(r):
    """Max drawdown of a return series, with peak / trough / recovery dates."""
    wealth = (1 + r.dropna()).cumprod()
    dd = wealth / wealth.cummax() - 1
    trough = dd.idxmin()
    peak = wealth[:trough].idxmax()
    after = wealth[trough:]
    rec = after[after >= wealth[peak]]
    return pd.Series({'Max Drawdown': dd.min(), 'Peak': peak, 'Trough': trough,
                      'Recovery': rec.index[0] if len(rec) else pd.NaT})

tail = pd.DataFrame({
    'Skewness':        hf.skew(),
    'Excess Kurtosis': hf.kurt(),              
    'VaR (.05)':       hf.quantile(.05),
    'CVaR (.05)':      pd.Series({c: hf[c][hf[c] <= hf[c].quantile(.05)].mean() for c in hf}),
})
tail = tail.join(pd.DataFrame({c: max_drawdown(hf[c]) for c in hf}).T)
tail  
```
