# 1. READING — The ProShares ETF Product

### 1.1 Alternative ETFs
> Describe the two types of investments referenced by this term.

"Alternative ETFs" covers two kinds of products:

1. **ETFs that give exposure to alternative asset classes**, i.e. assets outside the traditional long-only stock/bond universe, such as commodities, currencies, volatility, or real estate.
2. **ETFs that use alternative strategies**, i.e. hedge-fund-like techniques implemented inside an ETF wrapper: leveraged ("geared") and inverse funds, long/short and market-neutral funds, and hedge-fund replication products such as HDG.

The common thread is that they package exposures that historically required a hedge fund, a futures account, or derivatives into a liquid, exchange-traded, retail-accessible fund.

### 1.2 Hedge Funds
> a. Using just the information in the case, what are two measures by which hedge funds are an attractive investment?
>
> b. What are the main benefits of investing in hedge funds via an ETF instead of directly?

**a.** The case pitches hedge funds on two measures:

1. **Risk-adjusted return (Sharpe ratio).** Historically the HFRI earned returns comparable to equities but with substantially lower volatility, so its mean/volatility trade-off looked better than the stock market's.
2. **Low correlation / diversification.** Hedge-fund returns are less than perfectly correlated with stocks and bonds and held up comparatively well in equity drawdowns (e.g. 2000–02, 2008). Adding them to a traditional portfolio can therefore lower overall portfolio volatility.

(The case also notes the counterpoint that hedge funds had *underperformed* the stock market in the years just before 2014.)

**b.** Accessing hedge-fund exposure through an ETF like HDG offers:

* **Lower fees:** an expense ratio of about 0.95% versus the typical "2 and 20".
* **Liquidity:** shares trade daily on an exchange, with no lock-ups, redemption gates, or notice periods.
* **Accessibility:** no accredited-investor requirement and no large minimum investment; it can be bought in an ordinary brokerage account.
* **Transparency:** holdings are disclosed daily, unlike the opaque positions of most hedge funds.
* **No manager-selection risk or due diligence cost:** you get diversified "average hedge fund" exposure rather than betting on a single manager (and avoiding blow-up or fraud risk).
* **Regulation:** it is a registered fund, with simpler tax reporting than a private partnership.

### 1.3 The Benchmarks
> a. Explain as simply as possible how HFRI, MLFM, MLFM-ES, and HDG differ in their construction and purpose.
>
> b. How well does the Merrill Lynch Factor Model (MLFM) track the HFRI?
>
> c. In which factor does the MLFM have the largest loading? (See a slide in Exhibit 1.)
>
> d. What are the main concerns you have for how the MLFM attempts to replicate the HFRI?

**a.** Each is one step closer to something you can actually buy:

| | What it is | Purpose |
|---|---|---|
| **HFRI** (Fund Weighted Composite) | Equal-weighted average of the net-of-fee returns that a large universe of hedge funds self-report to HFR. | Benchmark for "the average hedge fund." **Not investable.** |
| **MLFM** | A model portfolio of six liquid factors (S&P 500, MSCI EAFE, MSCI EM, Russell 2000, EUR/USD via EUO, 3-month T-bills). Weights come from a rolling regression on trailing HFRI returns and are re-estimated monthly. | Replicate HFRI's returns (target high correlation) using only liquid, tradable assets. |
| **MLFM-ES** ("Exchange Series") | The same model, implemented with exchange-traded instruments (e.g. futures and ETFs) in place of the cash indexes. | An investable, rules-based index version of MLFM that a fund can actually hold. |
| **HDG** | ProShares ETF that holds those instruments plus T-bills. | Track MLFM-ES, net of a ~0.95% expense ratio, for retail investors. |

So the chain is: HFRI (target) → MLFM (factor replica) → MLFM-ES (tradable replica) → HDG (fund that delivers it).

**b.** MLFM tracks the HFRI closely in terms of co-movement, but not perfectly in level. In our Aug 2011 – Sep 2025 data (see the calculation cell below), the MLFM–HFRI correlation of monthly returns is about **0.90**, and annualized tracking error is about **2.6%**. However, MLFM's mean return is about **1.4% per year lower** than HFRI's. It captures HFRI's ups and downs well but does not reproduce its full return, which is consistent with HFRI containing some non-factor return (alpha and/or reporting biases) that six liquid factors cannot replicate.

**c.** Based on our full-sample regression of the MLFM on its six factors (see the calculation cell below), the largest loading is on **3-month T-bills (cash)**, at roughly 0.44. Among the risky factors, the largest are the **Russell 2000 (IWM)** and **MSCI EAFE (EFA)**, at roughly 0.13 each. (We did not have access to the Exhibit 1 slide, so this is our data-based estimate.) The heavy cash weight reflects the fact that hedge-fund returns have far lower volatility than equities, so the replica holds only a fraction in equities and parks the rest in T-bills.

**d.** Main concerns:

* **Backward-looking weights.** The rolling regression estimates exposures from past data, so it lags when hedge funds change their positioning, especially at turning points or in crises.
* **Linear, liquid-beta-only model.** Six equity/FX/cash factors cannot capture other sources of hedge-fund return: non-linear, option-like payoffs (e.g. short-volatility and tail risk), credit, illiquidity premia, merger-arbitrage spreads, or true manager alpha. The replica gets the "beta" piece only.
* **A flawed target.** HFRI is self-reported and suffers from survivorship, backfill, and selection bias. Its returns are also smoothed by illiquid, stale marks, which understates its volatility and its correlation with markets. Replicating HFRI therefore means chasing a number that overstates what hedge funds actually delivered.
* **Estimation error and overfitting.** Six correlated regressors fit over a short window give noisy betas. The weights can swing from month to month, generating turnover and trading costs, and in-sample fit overstates out-of-sample performance (see 2.7).
* **It replicates the average.** At best you get the average fund, after fees, not access to top managers.

### 1.4 The HDG Product
> a. What does ProShares ETF, HDG, attempt to track? Is the tracking error small?
>
> b. HDG is, by construction, delivering beta for investors. Isn't the point of hedge funds to generate alpha? Then why would HDG be valuable?
>
> c. The fees of a typical hedge-fund are 2% on total assets plus 20% of excess returns if positive. HDG's expense ratio is roughly 1% on total assets. What would their respective net Sharpe Ratios be, assuming both have a gross excess returns of 10% and volatility of 20%?

**a.** HDG tracks **MLFM-ES**, not HFRI directly. The ultimate goal is to resemble HFRI, but its stated benchmark is MLFM-ES. How small its tracking error is depends on which benchmark you compare it to:

* **Against its stated benchmark, MLFM-ES, tracking error is small.** The correlation is about 0.99 and annualized tracking error is about 0.9%. HDG's average shortfall is about 0.96% per year, which is almost exactly its 0.95% expense ratio, so it does what it promises.
* **Against HFRI, tracking error is larger.** The correlation is about 0.89 and annualized tracking error is about 2.7%. HDG's average return is about 2.5% per year below HFRI's. That shortfall stacks the MLFM's replication gap (about 1.4%) on top of fees and implementation costs (about 1%).

**b.** Several reasons:

* **Much of hedge-fund return *is* beta.** Factor exposures explain a large share of aggregate hedge-fund returns (the HFRI regression in 2.6 has R² ≈ 0.84). For the average fund, after-fee alpha is small or negative, so investors in hedge funds are largely paying 2/20 for exposures they could get much more cheaply.
* **The beta mix itself is useful.** It is a particular low-volatility combination of equity, international, small-cap, dollar, and cash exposure that diversifies a traditional portfolio. HDG delivers that mix with daily liquidity, transparency, and low fees.
* **No manager risk.** There is no fraud, blow-up, or style drift, and no lock-ups.
* **A natural benchmark.** HDG can serve as the "hedge-fund beta" against which investors measure whether active hedge-fund managers actually earn their fees. It can also be combined with a few genuine alpha managers.

**c.** Gross excess return is 10% and volatility is 20%, so the gross Sharpe ratio is 0.50.

* **Hedge fund:** management fee 2% plus performance fee 20% × 10% = 2%, for total fees of 4%. Net excess return = 6%, so **SR ≈ 6% / 20% = 0.30**.
* **HDG:** a fee of 1%, so net excess return = 9%, and **SR ≈ 9% / 20% = 0.45**.

So fees alone cut the hedge fund's Sharpe ratio by 40%, versus 10% for HDG.

*Robustness:* the answer depends somewhat on how the performance fee is modeled, but the conclusion does not change.

* If the 20% is charged on returns *after* the management fee and treated as proportional, net excess return = 10% − 2% − 0.2 × 8% = 6.4%. The fee also shrinks volatility to 0.8 × 20% = 16%, so SR = 0.40.
* If the incentive fee is charged only in positive years, with normally distributed returns (simulated below), SR ≈ 0.31. This version works like the manager holding a call option on your returns.

In every version the hedge fund's net Sharpe ratio is well below HDG's 0.45.

```python
# Supporting calculations for Section 1 (1.3b, 1.3c, 1.4a, 1.4c) -- self-contained
import numpy as np, pandas as pd, statsmodels.api as sm

_s  = pd.read_excel('data/proshares_analysis_data.xlsx', sheet_name=None, index_col=0)
_hf, _fac = _s['hedge_fund_series'], _s['merrill_factors']

def tracking(a, b):
    d = pd.concat([_hf[a], _hf[b]], axis=1).dropna()
    diff = d[a] - d[b]
    return pd.Series({'Correlation': d[a].corr(d[b]),
                      'Tracking error (ann.)': diff.std() * np.sqrt(12),
                      'Mean return gap (ann.)': diff.mean() * 12})

pairs = [('MLEIFCTR Index', 'HFRIFWI Index'),   # 1.3b  MLFM vs HFRI
         ('HDG US Equity',  'MLEIFCTX Index'),  # 1.4a  HDG vs its benchmark MLFM-ES
         ('HDG US Equity',  'HFRIFWI Index')]   # 1.4a  HDG vs HFRI
display(pd.DataFrame({f'{a} vs {b}': tracking(a, b) for a, b in pairs}).T)

# 1.3c cross-check: full-sample factor loadings of MLFM
_d = pd.concat([_hf['MLEIFCTR Index'], _fac], axis=1).dropna()
display(sm.OLS(_d['MLEIFCTR Index'], sm.add_constant(_d[_fac.columns])).fit().params.rename('MLFM loading').to_frame())

# 1.4c net Sharpe ratios
mu, sig = 0.10, 0.20
r = np.random.default_rng(0).normal(mu, sig, 1_000_000)
after_mgmt = r - 0.02
hf_asym = after_mgmt - 0.2 * np.maximum(after_mgmt, 0)       # incentive fee only when positive
display(pd.Series({
    'Gross':                               mu / sig,
    'Hedge fund (2% + 20% x 10%)':         (mu - 0.02 - 0.2 * mu) / sig,
    'Hedge fund (20% after mgmt fee, prop.)': (mu - 0.02 - 0.2 * (mu - 0.02)) / (0.8 * sig),
    'Hedge fund (incentive only if > 0, sim.)': hf_asym.mean() / hf_asym.std(),
    'HDG (1%)':                            (mu - 0.01) / sig,
}, name='Net Sharpe ratio').to_frame())
```

## 2.1 Summary Statistics
> For the series in the "hedge fund series" tab, report the following summary statistics: mean, volatility, Sharpe ratio. Annualize these statistics.

```python
# Series are treated as excess returns already, so Sharpe = mean / vol with no rf subtraction.
# HFRI has one NaN (Sep 2025), handled per column by pandas.
stats = pd.DataFrame({
    'Mean (ann.)': hf.mean() * FREQ,
    'Vol (ann.)':  hf.std() * np.sqrt(FREQ),
})
stats['Sharpe (ann.)'] = stats['Mean (ann.)'] / stats['Vol (ann.)']
stats
```

## 2.2 Tail Risk
> For the series in the "hedge fund series" tab, calculate the following statistics related to tail-risk.
> * Skewness
> * Excess Kurtosis (in excess of 3)
> * VaR (.05) - the fifth quantile of historic returns
> * CVaR (.05) - the mean of the returns at or below the fifth quantile
> * Maximum drawdown - include the dates of the max/min/recovery within the max drawdown period.
>
> There is no need to annualize any of these statistics.

```python
def max_drawdown(r):
    """Max drawdown of a return series, with peak / trough / recovery dates."""
    wealth = (1 + r.dropna()).cumprod()
    dd = wealth / wealth.cummax() - 1
    trough = dd.idxmin()
    peak = wealth[:trough].idxmax()
    after = wealth[trough:]
    rec = after[after >= wealth[peak]]
    return pd.Series({'Max Drawdown': dd.min(), 'Peak': peak, 'Trough': trough,
                      'Recovery': rec.index[0] if len(rec) else pd.NaT})

tail = pd.DataFrame({
    'Skewness':        hf.skew(),
    'Excess Kurtosis': hf.kurt(),              
    'VaR (.05)':       hf.quantile(.05),
    'CVaR (.05)':      pd.Series({c: hf[c][hf[c] <= hf[c].quantile(.05)].mean() for c in hf}),
})
tail = tail.join(pd.DataFrame({c: max_drawdown(hf[c]) for c in hf}).T)
tail  
```

## 2.3 Regression on SPY
> For the series in the "hedge fund series" tab, run a regression of each against SPY (found in the "merrill factors" tab.) Include an intercept. Report the following regression-based statistics:
> * Market Beta
> * Treynor Ratio
> * Information ratio
>
> Annualize these three statistics as appropriate.

```python
# 2.3 Regression on SPY
# Use excess returns on both sides and drop the one missing HFRI observation.
rf = factors['USGG3M Index']
market_ex = factors['SPY US Equity'] - rf
market_ex.name = 'SPY US Equity'

regression_results = []
for col in hf.columns:
    y = (hf[col] - rf).dropna()
    X = sm.add_constant(market_ex.loc[y.index])
    reg = sm.OLS(y, X).fit()

    beta = reg.params['SPY US Equity']
    treynor = (FREQ * y.mean()) / beta if abs(beta) > 1e-12 else np.nan
    info_ratio = (FREQ * reg.params['const']) / (reg.resid.std() * np.sqrt(FREQ))

    regression_results.append({
        'Series': col,
        'Market Beta': beta,
        'Treynor Ratio': treynor,
        'Information Ratio': info_ratio,
    })

regression_summary = pd.DataFrame(regression_results).set_index('Series')
regression_summary
```

## 2.4 Discussion
> Discuss the previous statistics, and what they tell us about…
> * the differences between SPY and the hedge-fund series?
> * which performs better between HDG and QAI.
> * whether HDG and the ML series capture the most notable properties of HFRI.



```python
# 2.4 support: put SPY on the same footing as the hedge-fund series (same definitions as 2.1-2.3)
spy     = factors['SPY US Equity']
spy_ex  = (spy - rf).rename('SPY US Equity')

spy_row = pd.DataFrame({
    'Mean (ann.)':     [spy_ex.mean() * FREQ],
    'Vol (ann.)':      [spy_ex.std() * np.sqrt(FREQ)],
    'Sharpe (ann.)':   [spy_ex.mean() / spy_ex.std() * np.sqrt(FREQ)],
    'Skewness':        [spy.skew()],
    'Excess Kurtosis': [spy.kurt()],
    'VaR (.05)':       [spy.quantile(.05)],
    'CVaR (.05)':      [spy[spy <= spy.quantile(.05)].mean()],
    'Max Drawdown':    [max_drawdown(spy)['Max Drawdown']],
    'Market Beta':     [1.0],
    'Treynor Ratio':   [spy_ex.mean() * FREQ],
    'Information Ratio': [np.nan],                 # alpha and residual are zero by construction
}, index=['SPY US Equity'])

compare = pd.concat([
    stats.join(tail[['Skewness', 'Excess Kurtosis', 'VaR (.05)', 'CVaR (.05)', 'Max Drawdown']])
         .join(regression_summary),
    spy_row,
]).astype(float)

compare.style.format({
    'Mean (ann.)': '{:.2%}', 'Vol (ann.)': '{:.2%}', 'Sharpe (ann.)': '{:.2f}',
    'Skewness': '{:.2f}', 'Excess Kurtosis': '{:.2f}', 'VaR (.05)': '{:.2%}', 'CVaR (.05)': '{:.2%}',
    'Max Drawdown': '{:.1%}', 'Market Beta': '{:.2f}', 'Treynor Ratio': '{:.3f}', 'Information Ratio': '{:.2f}',
}, na_rep='–')
```

The table above puts SPY next to the hedge-fund series, using the same definitions as 2.1–2.3 on this dataset.

**1. SPY vs. the hedge-fund series**

* **Risk and return.** SPY's mean excess return (≈ 13.0%) and volatility (≈ 14.3%) are both far higher than the hedge-fund series (means ≈ 1.2–3.6%, vols ≈ 5–6%). On a risk-adjusted basis SPY is also better: its Sharpe ratio of ≈ 0.90 beats every hedge-fund series (0.21–0.62). In this sample, lower risk did not come with a better risk–return trade-off.
* **Market exposure.** The hedge-fund series have market betas of only ≈ 0.30–0.35, i.e. about a third of SPY's market exposure. Their Treynor ratios (0.03–0.11) are all below SPY's 0.13, and all their information ratios are **negative**. After accounting for market beta, none of them delivered positive alpha relative to SPY. HFRI comes closest (IR ≈ −0.26).
* **Tail risk.** In absolute terms the hedge-fund series have smaller losses: VaR and CVaR are less than half of SPY's, and max drawdowns are ≈ 11–14% vs. ≈ 24% for SPY. This mostly reflects their lower volatility. The *shape* of the distribution tells a different story. HFRI is much more negatively skewed (−0.95 vs. −0.41) and has much fatter tails (excess kurtosis 5.7 vs. 0.7). Relative to its volatility, HFRI's crash risk is worse than SPY's.

**2. HDG vs. QAI**

QAI did better on almost every metric. It has a higher mean (1.37% vs. 1.18%), lower volatility (4.9% vs. 5.7%), and a higher Sharpe ratio (0.28 vs. 0.21). Its VaR (−1.7% vs. −3.0%) and CVaR (−3.1% vs. −3.7%) are smaller, and its max drawdown is slightly smaller (−13.8% vs. −14.1%), with a faster recovery (Feb 2024 vs. Jul 2024). It also has a lower beta, a higher Treynor ratio, a less negative information ratio, and lower kurtosis. The one exception is **skewness**, where HDG (−0.27) is less negative than QAI (−0.43).

These differences are small, though. With about 14 years of data, the standard error of an annualized Sharpe ratio is roughly $1/\sqrt{14} \approx 0.27$, much larger than the 0.07 gap. So we cannot say that QAI is reliably the better product.

**3. Do HDG and the ML series capture the most notable properties of HFRI?**

* **What they capture:** co-movement (correlations with HFRI ≈ 0.89–0.90, see 2.5), the level of volatility (≈ 5.5–5.7% vs. 5.9%), and market beta (≈ 0.34–0.35 vs. 0.35). They track HFRI's month-to-month swings and its equity exposure well.
* **What they miss:**
  * **The return level.** Their means are 1.3–2.5% per year below HFRI's, and their Sharpe ratios (0.21–0.43) are well below HFRI's 0.62. Part of this gap is the intercept in 2.6 (≈ 1.4% per year), which no combination of the factors can reproduce. HDG gives up roughly another 1% per year in fees.
  * **The tails.** HFRI's most distinctive feature is its strong negative skew (−0.95) and fat tails (excess kurtosis 5.7). The replicators show only mild skew (≈ −0.28) and kurtosis (≈ 1.6–1.8). A linear combination of liquid index returns cannot reproduce the non-linear, option-like payoffs that produce HFRI's crash risk. The max-drawdown timing differs too. HFRI's worst drawdown was the COVID crash (Dec 2019 – Mar 2020, recovered by Aug 2020). For all the replicators it was the 2021–22 sell-off (Jun 2021 – Sep 2022, recovered only in 2024).

So HDG and the ML series capture HFRI's *beta and volatility*, but not its *return premium* or its *crash tail*. Some of HFRI's apparent premium may itself come from self-reporting and survivorship bias in the index (see 1.3d), so this part may not be worth chasing.

## 2.6 Replicating HFRI (in-sample)
> Replicate HFRI with the six factors listed on the "merrill factors" tab. Include a constant, and run the unrestricted regression,
>
> $$r^{HFRI}_{t} = \alpha^{merr} + x_{t}^{merr}\beta^{merr} + \epsilon_{t}^{merr}$$
> $$\hat{r}^{HFRI}_{t} = \hat{\alpha}^{merr} + x_{t}^{merr}\hat{\beta}^{merr}$$
>
> a. Report the intercept and betas.
>
> b. Are the betas realistic position sizes, or do they require huge long-short positions?
>
> c. Report the R-squared.
>
> d. Report the volatility of $\epsilon^{merr}$, the tracking error.

```python
# 2.6 Replicating HFRI (in-sample)
factor_cols = ['SPY US Equity', 'USGG3M Index', 'EEM US Equity', 'EFA US Equity', 'EUO US Equity', 'IWM US Equity']
reg_df = pd.concat([hf['HFRIFWI Index'], factors[factor_cols]], axis=1).dropna()

X = sm.add_constant(reg_df[factor_cols])
y = reg_df['HFRIFWI Index']

hfri_reg = sm.OLS(y, X).fit()

beta_summary = pd.DataFrame({
    'Factor': factor_cols,
    'Beta': hfri_reg.params[factor_cols].values
})

print('Intercept:', hfri_reg.params['const'])
print('Betas:', hfri_reg.params[factor_cols].to_dict())
print('R-squared:', hfri_reg.rsquared)
print('Tracking error (monthly):', hfri_reg.resid.std())
print('Tracking error (annualized):', hfri_reg.resid.std() * np.sqrt(FREQ))
```

```python
display(pd.concat([
    hfri_reg.params.rename('Estimate'),
    hfri_reg.tvalues.rename('t-stat'),
], axis=1).style.format('{:.4f}'))
print(f'Sum of factor betas: {hfri_reg.params[factor_cols].sum():.3f}')
print(f'Intercept (annualized): {hfri_reg.params["const"] * FREQ:.2%}')
```

**Takeaways (2.6)**

**a. Intercept and betas** (table above). The intercept is ≈ 0.11% per month (≈ 1.4% per year, t ≈ 1.6). The betas are SPY 0.04, T-bills 0.32, EEM 0.09, EFA 0.07, EUO 0.03, and IWM 0.15.

**b. Realistic position sizes?** Yes. All six betas are **positive and small**: the largest is 0.32 (T-bills), and the sum of the betas is ≈ 0.70. The risky equity and FX positions add up to only ≈ 0.38, which is consistent with HFRI's market beta of ≈ 0.35 in 2.3. The replication needs **no large long-short positions** and could be implemented long-only, which is very different from the extreme weights we saw in mean-variance optimization in HW1.

Individual betas are imprecise, though. IWM (t ≈ 7.6), EEM (t ≈ 4.5), and EFA (t ≈ 2.4) are clearly significant. SPY is not (t ≈ 1.4) because it overlaps heavily with IWM and EFA. The T-bill beta is very poorly identified (t ≈ 1.0): T-bill returns were close to zero and barely moved for most of the sample, so they are almost collinear with the constant. 2.7 shows how unstable this beta becomes in rolling windows.

**c. R-squared:** ≈ **0.84**. The six factors explain about 84% of HFRI's monthly variation.

**d. Tracking error:** the volatility of $\epsilon^{merr}$ is ≈ 0.67% per month, or ≈ **2.3% annualized**. That is substantial relative to HFRI's own volatility of ≈ 5.9%. Note also that the 1.4% per-year intercept is part of the fitted value $\hat{r}^{HFRI}_t$ but cannot be bought: a tradeable replica holds only the factor positions, so it would trail HFRI by roughly that amount on average.

## 2.7 Out-of-Sample Replication
> Starting with $t = 61$ month of the sample, use the previous 60 months of data to estimate the regression equation. This gives time-$t$ estimates $\tilde{\alpha}^{merr}_{t}$ and $\tilde{\beta}^{merr}_{t}$. Use them with the time-$t$ regressor values to calculate the OOS replication value
>
> $$\hat{r}^{HFRI}_{t} \equiv \tilde{\alpha}^{merr} + (x_{t}^{merr})'\tilde{\beta}^{merr}$$
>
> Step forward one month at a time (rolling 60-month regression).
>
> How well does the out-of-sample replication perform with respect to the target?

```python
# 2.7 Rolling 60-month out-of-sample replication
WINDOW = 60
X_all = sm.add_constant(reg_df[factor_cols])       # reg_df from 2.6 (HFRI + 6 factors, NaN month dropped)
y_all = reg_df['HFRIFWI Index']

oos_pred, roll_params = {}, {}
for t in range(WINDOW, len(reg_df)):                # t = 61st month onward (0-indexed position 60)
    fit_t = sm.OLS(y_all.iloc[t - WINDOW:t], X_all.iloc[t - WINDOW:t]).fit()   # months t-60 ... t-1 only
    date  = reg_df.index[t]
    oos_pred[date]    = float(X_all.iloc[t] @ fit_t.params)                     # alpha_t + x_t' beta_t
    roll_params[date] = fit_t.params

oos_rep     = pd.Series(oos_pred, name='OOS replication')
roll_params = pd.DataFrame(roll_params).T
target      = y_all.loc[oos_rep.index].rename('HFRI')
is_rep      = hfri_reg.fittedvalues.loc[oos_rep.index].rename('In-sample replication')   # 2.6 fit, same months

print(f'OOS period: {oos_rep.index[0]:%Y-%m} to {oos_rep.index[-1]:%Y-%m}  ({len(oos_rep)} months)')

def replication_stats(rep, tgt=target):
    err = tgt - rep
    return pd.Series({
        'Correlation with HFRI':    rep.corr(tgt),
        'R-squared (1 - SSE/SST)':  1 - (err ** 2).sum() / ((tgt - tgt.mean()) ** 2).sum(),
        'Tracking error (ann.)':    err.std() * np.sqrt(FREQ),
        'Mean (ann.)':              rep.mean() * FREQ,
        'Vol (ann.)':               rep.std() * np.sqrt(FREQ),
    })

rep_stats = pd.DataFrame({'OOS replication': replication_stats(oos_rep),
                          'In-sample replication': replication_stats(is_rep)})
rep_stats.loc['Mean (ann.)', 'HFRI'] = target.mean() * FREQ
rep_stats.loc['Vol (ann.)', 'HFRI']  = target.std() * np.sqrt(FREQ)
rep_stats.style.format('{:.4f}', na_rep='–')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

(1 + pd.concat([target, oos_rep, is_rep], axis=1)).cumprod().plot(ax=axes[0])
axes[0].set_title('Growth of $1: HFRI vs. replications (OOS period)')
axes[0].set_ylabel('Value of $1')

# T-bill beta omitted from the plot: nearly collinear with the constant, so it swings wildly (see table below)
roll_params[[c for c in factor_cols if c != 'USGG3M Index']].plot(ax=axes[1])
axes[1].axhline(0, color='k', lw=0.8)
axes[1].set_title('Rolling 60-month betas used for the OOS replication')
axes[1].legend(fontsize=8)

plt.tight_layout()
plt.show()

roll_params.describe().loc[['mean', 'std', 'min', 'max']].style.format('{:.3f}')
```

**Takeaways (2.7)**

* **Setup.** The first OOS month is Aug 2016, the 61st observation of a sample that starts in Aug 2011. The last is Aug 2025: Sep 2025 is dropped because HFRI is missing that month. That gives 109 OOS months. Each month's replication uses only the *previous* 60 months to estimate $\tilde{\alpha}_t$ and $\tilde{\beta}_t$, so there is no look-ahead.
* **Co-movement holds up well out of sample.** The OOS replication has a correlation of ≈ 0.90 with HFRI, an OOS $R^2$ (1 − SSE/SST) of ≈ 0.81, and a tracking error of ≈ 2.8% per year. Over the same months, the full-sample (in-sample) fit gives 0.92, 0.85, and 2.5%. Going out of sample costs only a little, which suggests HFRI's factor exposures are fairly stable over time. The result is also about as good as the actual product: in Section 1, HDG vs. HFRI had a correlation of ≈ 0.89 and a tracking error of ≈ 2.7%.
* **The return level does not hold up.** Over the OOS period the replication averaged ≈ 4.8% per year vs. ≈ 6.4% for HFRI, a shortfall of about 1.6% per year. $1 grew to about $1.52 vs. $1.76 for HFRI. The in-sample fit closes only part of this gap (≈ 5.7%). This OOS replication also *includes* the estimated rolling intercept, which an investor cannot buy, so a tradeable replica holding only the factor positions would fall even further behind.
* **The betas are not stable.** The equity betas move around noticeably. IWM ranges from ≈ 0.05 to 0.25, and SPY ranges from ≈ −0.05 to 0.12, sometimes turning negative. The T-bill beta is extremely unstable, ranging from about −2.5 to 10.2, so it is left out of the plot. With T-bill returns near zero and flat for most of 2011–2021, it is nearly collinear with the intercept, and the regression assigns it huge and meaningless weights. In practice, swings like these would generate turnover and trading costs. That is one reason Merrill constrains the weights in its model (see EXTRA 3.2).

**Conclusion:** the rolling OOS replication tracks HFRI's monthly ups and downs almost as well as the in-sample fit does, but it does not deliver HFRI's average return. This is consistent with 2.4: factor replication captures hedge funds' beta, not their (reported) premium.
